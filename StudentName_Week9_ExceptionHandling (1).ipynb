## Task 1 – Identify the errors
Run the three examples provided below and observe the errors.

### Example 2

```python
# Example 2
len(5)
```

**Error type:** `TypeError`

**Explanation:** This error occurs because the built-in function `len()` only works on objects that have a length, such as strings, lists, tuples, or dictionaries. An integer (`5`) does not have a length, so Python raises a `TypeError: object of type 'int' has no len()`.

### Example 3

```python
# Example 3
def add(a, b):
    return a + b

add('hello', 5)
```

**Error type:** `TypeError`

**Explanation:** This error occurs because the function tries to add a string (`'hello'`) and an integer (`5`) together using the `+` operator. Python does not know how to concatenate a `str` and an `int`, so it raises `TypeError: can only concatenate str (not "int") to str`.

### Example 4 (OverFlowError)

```python
#OverFlowError
import math
x = math.exp(1000)
```

**Error type:** `OverflowError`

**Explanation:** This error occurs because `math.exp(1000)` tries to calculate e¹⁰⁰⁰, a number far too large to be represented as a Python float. Since the result exceeds the maximum size a floating-point number can hold, Python raises an `OverflowError: math range error`.

## Task 2 & 3 – Handle the errors and test your solution
Each program below has been modified with a `try...except` block that catches the correct exception, prevents the program from crashing, and displays a clear, meaningful message. Each cell is run to show the corrected output.

### Example 2 – Fixed (TypeError)

```python
# Example 2 - with try...except
try:
    print(len(5))
except TypeError:
    print("Error: Invalid input. Please enter the correct type of value (e.g. a string, list, or tuple).")
```

### Example 3 – Fixed (TypeError)

```python
# Example 3 - with try...except
def add(a, b):
    return a + b

try:
    result = add('hello', 5)
    print("Result:", result)
except TypeError:
    print("Error: Invalid input. Please make sure both values are the same type (e.g. two numbers or two strings).")
```

### Example 4 (OverFlowError) – Fixed

```python
# Example 4 - with try...except
import math

try:
    x = math.exp(1000)
    print("Result:", x)
except OverflowError:
    print("Error: The calculation result is too large to compute. Please try a smaller number.")
```

## Summary

| Example | Error Type | Cause |

| 3 | `TypeError` | Adding a string and an integer together |

Each program now catches its specific exception with a `try...except` block, preventing the program from stopping unexpectedly and instead showing the user a clear, friendly message.
